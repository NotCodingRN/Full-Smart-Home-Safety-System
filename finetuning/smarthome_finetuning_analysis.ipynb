# Smart Home ML and Telemetry Analysis
This notebook documents FaceNet embedding threshold tuning and representative sensor/control statistics.

```python
import csv, statistics
rows=list(csv.DictReader(open('../data/demo_sensor_control_log.csv')))
print('samples', len(rows))
print('mq2 max', max(int(r['mq2_adc']) for r in rows))
print('mean inference ms', round(statistics.mean(float(r['inference_ms']) for r in rows), 2))
```
